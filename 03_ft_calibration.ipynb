# FT 学习率校准：共享仓库
先选择项目 .venv 内核。8 条件 × 3 学习率 × 3 折 = 72 次 inner fits；无 outer test。完整运行可能需要较长时间，支持恢复已完成 fits。

```python
from pathlib import Path
import subprocess, sys
ROOT = Path.cwd()
REPO = ROOT if (ROOT / "calibrate_ft.py").exists() else ROOT / "sml-code"
assert (REPO / "calibrate_ft.py").exists()
subprocess.run([sys.executable, "calibrate_ft.py"], cwd=REPO, check=True)
```

```python
RUN_CALIBRATION = False
if RUN_CALIBRATION:
    process = subprocess.Popen([sys.executable, "-u", "calibrate_ft.py", "--execute"], cwd=REPO, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, encoding="utf-8", errors="replace")
    try:
        for line in process.stdout:
            print(line, end="", flush=True)
        if process.wait() != 0:
            raise RuntimeError("校准失败，请查看上面的错误")
    except BaseException:
        process.terminate()
        process.wait()
        raise
else:
    print("将 RUN_CALIBRATION 改为 True 开始运行。")
```

```python
import json
path = REPO / "analysis/ft_calibration_v4/calibration.json"
if path.exists():
    results = json.loads(path.read_text(encoding="utf-8"))
    print(results["status"])
    for selection in results["selections"]:
        print(selection["condition"], "selected LR:", selection["value"])
else:
    print("尚无完整条件结果。")
```
